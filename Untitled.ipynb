```python
import os
import re
import zipfile
from pathlib import Path
import requests
from requests.exceptions import RequestException

# Use a list of your file URIs
URIS = [
    "https://example.com/api/download?file_id=123", # URI with no filename
    "https://example.com/downloads/report.zip", # URI with filename
    "https://example.com/invalid_link", # Invalid URI
]

def get_filename_from_cd(content_disposition):
    """
    Extracts filename from the Content-Disposition header.
    Returns None if the header is not found or is invalid.
    """
    if content_disposition:
        fname = re.findall("filename=(.+)", content_disposition)
        if len(fname) > 0:
            # Clean up the filename from quotes
            return fname[0].strip('"')
    return None

def download_and_extract(uri, download_dir="downloads"):
    """
    Downloads a single file, extracts its contents, and removes the zip.
    """
    try:
        with requests.get(uri, stream=True, allow_redirects=True, timeout=30) as r:
            r.raise_for_status()  # Raise HTTPError for bad responses (4xx or 5xx)

            # Determine filename from Content-Disposition header first
            filename = get_filename_from_cd(r.headers.get("Content-Disposition"))
            if not filename:
                # Fallback to URL path if Content-Disposition is not present
                filename = Path(r.url).name
                if not filename:
                    # Generic filename if all else fails
                    filename = f"file_{uri.split('?')[0].split('/')[-1]}.zip"
            
            filepath = os.path.join(download_dir, filename)

            # Download the zip file
            print(f"Downloading {filename} from {uri}...")
            with open(filepath, "wb") as f:
                for chunk in r.iter_content(chunk_size=8192):
                    f.write(chunk)
            
            # Extract the CSV from the zip file
            print(f"Extracting CSV from {filename}...")
            with zipfile.ZipFile(filepath, 'r') as zip_ref:
                for member in zip_ref.namelist():
                    # Check if the file is a CSV and not in a sub-folder
                    if member.endswith('.csv') and '/' not in member:
                        zip_ref.extract(member, download_dir)
                        print(f"Extracted {member} to {download_dir}")
            
            # Delete the zip file
            os.remove(filepath)
            print(f"Deleted {filename}.")

    except RequestException as e:
        print(f"Error downloading {uri}: {e}")
    except zipfile.BadZipFile:
        print(f"Error: Downloaded file from {uri} is not a valid zip file.")
    except Exception as e:
        print(f"An unexpected error occurred for {uri}: {e}")

def main(uris):
    download_dir = "downloads"
    os.makedirs(download_dir, exist_ok=True)
    for uri in uris:
        download_and_extract(uri, download_dir)

if __name__ == "__main__":
    main(download_uris)
```
